# Lecture - Planning the Project

## Overview

In this lecture, you’ll learn how to plan a data visualization project by structuring environments, managing data across dev, stage, and prod, and setting up a CI/CD pipeline in Databricks for smooth, secure deployments. 

## Learning Objectives

By the end of this lecture, you will be able to: 
1. Learn to plan and structure a data engineering project, defining key components and isolation steps for successful execution.

##### EXPAND FOR ADDITIONAL NOTES
<details>
<p>Within a CI/CD pipeline</p>
<p>In development, data is often anonymized or generated using synthetic datasets to allow rapid development and testing without compromising privacy or production data integrity.</p>
<p>If you have an staging environment, staging data should closely mirror production in structure and volume, with anonymized or scrubbed sensitive information to ensure realistic testing and validation.</p>
<p>Production data is live, fully operational, and continuously updated, containing real user data, and must be handled with high security, privacy, and compliance standards.</p>
<p>Each environment should have data suited for its specific purpose, balancing realism, security, and compliance at every stage. How this is implemented will depend on your organization and the sensitivity of your data.</p>
</details>

##### EXPAND FOR ADDITIONAL NOTES
<details>
<p>Now that we have an understanding of our project, our data, and how to isolate environments, let&#x27;s dive into the project setup.</p>
<p>In this project we will create a catalog for each environment: dev, stage, and prod.</p>
<p>The dev catalog will contain a small, static subset of our production data, used for development and initial testing.</p>
<p>The stage catalog will hold a larger subset of production data, enabling more comprehensive testing as we move through our CI/CD process.</p>
<p>Finally, the prod catalog will contain the live production data that we rely on for final operations.</p>
<p>Our workflow starts by developing on the dev data, running unit and integration tests as we progress through the pipeline. The pipeline is set up with Databricks Lakeflow Jobs, which execute the unit tests, Spark Declarative pipeline, and final visualizations.</p>
<p>As we test the pipeline through each stage, we will ensure everything is functioning correctly before deploying to production.</p>
</details>
