# 5.1 Lecture: Lakehouse Federation Overview

This lesson introduces Lakehouse Federation in Unity Catalog and draws a clear distinction between its two flavours: catalog federation, which mirrors a legacy metastore into UC, and lakehouse federation, which exposes an external operational or analytical database in place. The rest of Section 5 walks through both forms with hands-on demos and a lab.

## Learning Objectives

By the end of this lesson, you will be able to:
- Define Lakehouse Federation as implemented by Unity Catalog
- Distinguish catalog federation from lakehouse federation
- Identify the components shared by both forms (connections and foreign catalogs)
- Match each form to the use cases it solves
- Recognize the supported source systems for each form

## A. What is Lakehouse Federation?

Lakehouse Federation is an umbrella term for the UC features that let you query data living outside Unity Catalog as if it were a UC asset, without copying or moving the data.

Both forms of federation use the same two building blocks:
- A **Connection** that holds credentials and configuration for the external source
- A **Foreign Catalog** that mirrors the source's namespace into UC

## B. Catalog Federation vs lakehouse Federation

The two forms of federation differ in *what* is being federated. Catalog federation points at another *metastore*; lakehouse federation points at another *database*.

## C. Side-by-Side Comparison

The table below summarizes how the two forms differ in source, primary use case, and where each is covered later in this section.

| Dimension | Catalog Federation | Lakehouse Federation |
|---|---|---|
| **What is federated** | An external metastore | An external database |
| **Sources** | Hive Metastore, AWS Glue | PostgreSQL (incl. Lakebase), SQL Server, Snowflake, MySQL, Redshift, BigQuery, others |
| **Tables exposed** | Existing tables in the source metastore | Tables/views in the source database |
| **Read/Write** | Read in UC; write through source engine | Read in UC; some sources support write |
| **Primary use case** | Migrating from legacy metastore to UC | Querying OLTP / external warehouse data alongside lakehouse data |
| **Long-term direction** | Bridge to UC managed tables | Long-lived federation; data stays in source |
| **Covered in** | 5.3 (lecture), 5.4 (demo) | 5.2 (demo), 5.5 (lab on Lakebase) |

## D. Catalog Federation: Bridging Legacy Metastores

Catalog federation is most often used to migrate off a legacy Hive Metastore or AWS Glue catalog into Unity Catalog without disrupting the workloads that already run against those tables.

**Typical adoption path:**
1. Register the legacy metastore as a foreign catalog - existing tables become queryable through UC immediately
2. Workloads gradually shift to read through UC, picking up UC governance
3. Tables are migrated in-place from foreign to external to managed
4. Once all tables are migrated, the foreign catalog can be retired

Covered in detail in [5.3 Lecture - Migration to UC using a Federated Catalog]($./5.3 Lecture - Migration to UC using a Federated Catalog) with the AWS Glue walkthrough in [5.4 Demo - Creating a Foreign Catalog Connection to AWS Glue Catalog]($./5.4 Demo - Creating a Foreign Catalog Connection to AWS Glue Catalog).

## E. Lakehouse Federation: Joining External Databases

Lakehouse federation exposes a live external database through Unity Catalog so that you can query and join its data in place, with supported predicates, projections, and aggregations pushed down to the source.

**Typical use cases:**
- Joining operational data in PostgreSQL, MySQL, or SQL Server with analytical data in the lakehouse without ETL
- Reading from another data warehouse (Snowflake, Redshift, BigQuery) under UC governance
- Exposing OLTP data sitting in **Lakebase** (Databricks-managed PostgreSQL) to lakehouse workloads

Demonstrated against SQL Server and Snowflake in [5.2 Demo - Federation to SQL Server and Snowflake]($./5.2 Demo - Federation to SQL Server and Snowflake) and exercised hands-on against Lakebase in [5.5 Lab - Implementing Lakehouse Federation]($./5.5 Lab - Implementing Lakehouse Federation).

## F. Pushdown Behavior

For Lakehouse federation, performance depends heavily on what UC can push down to the source rather than pulling rows back to Databricks for processing.

| Pushdown | Typically Supported |
|---|---|
| Filters, projections, limits | All sources |
| Aggregates and arithmetic operators | Most sources, on DBR 13.3 LTS+ / SQL Warehouses |
| Sorting (with limit) | Most sources, on DBR 13.3 LTS+ / SQL Warehouses |
| Joins | Newer DBR (e.g. PostgreSQL needs DBR 17.2+ / SQL Warehouses 2025.30+) |
| Window functions | Generally not pushed down |

Specifics vary by source; the lab in 5.5 walks through reading the `EXPLAIN` output to confirm what was pushed down for a federated PostgreSQL query.

## Key Takeaways

- **Lakehouse Federation** covers two distinct patterns built on the same connection + foreign catalog primitives
- **Catalog federation** points UC at another metastore (HMS, Glue) - typically used as a migration bridge
- **Lakehouse federation** points UC at another database (PostgreSQL/Lakebase, SQL Server, Snowflake, etc.) - typically used as a long-lived integration
- Both keep data in place; **credentials and configuration** live in the connection, the namespace lives in the foreign catalog
- **Pushdown** is what determines federation performance for lakehouse federation - check `EXPLAIN` to verify
