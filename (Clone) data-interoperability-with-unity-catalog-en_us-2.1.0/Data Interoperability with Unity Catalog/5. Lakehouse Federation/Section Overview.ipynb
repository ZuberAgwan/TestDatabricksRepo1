# 5. Lakehouse Federation - Overview

This section covers Lakehouse Federation in two forms: **Lakehouse federation** to external databases (PostgreSQL/Lakebase, SQL Server, Snowflake) and **catalog federation** for migrating from legacy metastores like AWS Glue into Unity Catalog. The instructor demos Lakehouse federation against SQL Server and Snowflake, learners exercise the same pattern hands-on against Lakebase, and a separate lecture/demo pair covers catalog federation as a migration bridge.

## Learning Objectives

By the end of this section, you will be able to:
- Distinguish catalog federation from Lakehouse federation
- Create connections and foreign catalogs for either form
- Run federated queries against external databases and metastores
- Identify supported pushdowns and verify behavior with `EXPLAIN`
- Plan a phased migration from a legacy metastore to UC managed tables
