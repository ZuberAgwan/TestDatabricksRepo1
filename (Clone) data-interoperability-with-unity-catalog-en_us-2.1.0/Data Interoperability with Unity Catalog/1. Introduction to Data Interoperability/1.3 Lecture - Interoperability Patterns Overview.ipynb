# 1.3 Lecture: Interoperability Patterns Overview

This lesson surveys the main interoperability patterns supported by Unity Catalog - the Iceberg REST Catalog, catalog federation, and lakehouse federation - and notes where Open Sharing fits as a related but separately scoped capability. Each pattern is covered in depth in its own dedicated section later in the course.

## Learning Objectives

By the end of this lesson, you will be able to:
- Identify Unity Catalog as the central governance layer across patterns
- Distinguish the three primary interoperability patterns: Iceberg REST, catalog federation, and lakehouse federation
- Recognize where Open Sharing fits at a high level
- Map each pattern to the section of the course where it is covered in detail

## A. Unity Catalog: Central Governance Layer

Unity Catalog sits above the storage layer and applies a single set of governance controls regardless of which table format the data is stored in or which engine is querying it.

**When to use it:**
- Downstream analytics engines (Snowflake, Trino, Dremio, Starburst) need access to UC tables
- External Spark, Flink, or PyIceberg clients perform compute outside Databricks against UC-governed data

## Key Takeaways

- **Unity Catalog** is the central governance layer regardless of which interoperability pattern is in play
- The **Iceberg REST Catalog** connects external Iceberg engines to UC managed tables
- **Catalog federation** bridges legacy metastores (HMS, Glue) into UC
- **Lakehouse federation** exposes external databases (PostgreSQL/Lakebase, SQL Server, Snowflake) through UC
- **Open Sharing** is the right tool for cross-organization data sharing - mentioned here, but out of scope for this course
