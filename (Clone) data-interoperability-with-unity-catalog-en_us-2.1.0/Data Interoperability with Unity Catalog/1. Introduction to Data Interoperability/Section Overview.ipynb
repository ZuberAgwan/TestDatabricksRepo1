# 1. Introduction to Data Interoperability - Overview

This section introduces the fundamental concepts of data interoperability in the modern data ecosystem, takes a closer look at the open table formats that underpin the lakehouse, and surveys the main interoperability patterns supported by Unity Catalog.

## Learning Objectives

By the end of this section, you will be able to:
- Define Data Interoperability and its business value
- Identify common data interoperability challenges
- Compare the anatomy and capabilities of Delta Lake and Apache Iceberg
- Explain how UniForm bridges Delta and Iceberg
- Identify the main interoperability patterns: Iceberg REST, catalog federation, Lakehouse federation
- Recognize where Open Sharing fits at a high level
