```sql
-- Preflight: ensure required tables exist for this module
BEGIN
  DECLARE schema_exists BOOLEAN DEFAULT FALSE;
  DECLARE table_exists BOOLEAN DEFAULT FALSE;

  SET schema_exists = (
    SELECT COUNT(*) > 0
    FROM system.information_schema.schemata
    WHERE catalog_name = current_catalog()
      AND schema_name = 'data_interoperability_tpcds'
  );

  IF NOT schema_exists THEN
    SELECT raise_error(
      'Schema "' || current_catalog() || '.data_interoperability_tpcds" not found. ' ||
      'Run the "0 - Required Setup" notebook before running this notebook.'
    );
  END IF;

  SET table_exists = (
    SELECT COUNT(*) > 0
    FROM system.information_schema.tables
    WHERE table_catalog = current_catalog()
      AND table_schema = 'data_interoperability_tpcds'
      AND table_name = 'store_sales_unclustered'
  );

  IF NOT table_exists THEN
    SELECT raise_error(
      'Required table "' || current_catalog() || '.data_interoperability_tpcds.store_sales_unclustered" not found. ' ||
      'Run the "1 - Instructor Demo Setup" notebook before running this notebook.'
    );
  END IF;

  SET table_exists = (
    SELECT COUNT(*) > 0
    FROM system.information_schema.tables
    WHERE table_catalog = current_catalog()
      AND table_schema = 'data_interoperability_tpcds'
      AND table_name = 'store_sales_clustered'
  );

  IF NOT table_exists THEN
    SELECT raise_error(
      'Required table "' || current_catalog() || '.data_interoperability_tpcds.store_sales_clustered" not found. ' ||
      'Run the "1 - Instructor Demo Setup" notebook before running this notebook.'
    );
  END IF;
END;
```
