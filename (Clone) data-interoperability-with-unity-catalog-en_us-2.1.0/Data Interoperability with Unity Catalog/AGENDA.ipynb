# Data Interoperability with Unity Catalog

## Course Overview

This course explores how to achieve data interoperability between systems using Databricks Unity Catalog as a unified governance layer, using both Delta and Iceberg open table formats. Participants will learn to implement, manage and optimize a data environment where tables can be accessed by both Delta and Iceberg clients, enabling cross-platform analytics and eliminating data silos.

This course is designed for **Data Engineers**, **Platform Administrators**, and **Analytics Engineers** responsible for implementing and managing multi-format data architectures. You will gain the skills to use Unity Catalog for unified governance across Delta and Iceberg table formats and across federated external systems.

## Table of Contents

A quick map of what this agenda covers and where to jump.

| # | Section | Summary |
|---|---------|---------|
| 1 | **Introduction to Data Interoperability** | Concepts, open table formats (Delta, Iceberg, UniForm), and the interoperability patterns UC supports |
| 2 | **Working with Managed Tables in Unity Catalog** | UC governance, Delta vs Iceberg managed tables, performance optimizations, and external access setup |
| 3 | **Centralized Data Processing with External Analytics** | Databricks as the processing hub; external engines like Snowflake and PyIceberg consuming UC via the Iceberg REST endpoint |
| 4 | **Centralized Governance with UC with External Compute Engines** | UC as the governance authority while external compute (e.g. AWS EMR) reads and writes managed tables |
| 5 | **Lakehouse Federation** | Catalog vs lakehouse federation; demos against SQL Server, Snowflake, AWS Glue; Lakebase lab |
| 6 | **Best Practices** | Decision framework, governance principles, and common pitfalls across interoperability projects |

## Terminal Learning Objective

**Implement Effective Data Interoperability Strategies Using Unity Catalog**

By the end of this course, you will be able to:

- Understand data interoperability and its importance in modern data ecosystems
- Compare Delta and Iceberg open table formats and use UniForm to bridge them
- Identify the main interoperability patterns supported by Unity Catalog (Iceberg REST, catalog federation, lakehouse federation)
- Use Unity Catalog as the central governance layer across systems
- Implement and manage Delta and Iceberg managed tables in Databricks
- Use the Iceberg REST API for accessing UC tables from external engines (Snowflake, PyIceberg, EMR)
- Implement a medallion architecture pipeline that works across Delta and Iceberg
- Configure security and access controls for cross-platform data sharing
- Enable external compute engines to read and write UC managed tables under UC governance
- Configure Lakehouse Federation to PostgreSQL/Lakebase, SQL Server, Snowflake, and AWS Glue
- Apply governance, architecture, and implementation principles for cross-platform interoperability
