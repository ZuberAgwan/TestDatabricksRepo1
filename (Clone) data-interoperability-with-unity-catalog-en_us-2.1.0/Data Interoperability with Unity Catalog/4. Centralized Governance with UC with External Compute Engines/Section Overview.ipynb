# 4. Centralized Governance with UC with External Compute Engines - Overview

This section covers the second major interoperability use case: external compute engines like AWS EMR or Apache Flink running transformations and writing back to UC managed Iceberg tables, with Unity Catalog continuing to enforce governance over the data they touch. The lecture introduces the pattern and the demo walks through the EMR-side configuration.

## Learning Objectives

By the end of this section, you will be able to:
- Describe the centralized governance pattern with external compute
- Identify the privileges required for external read/write access
- Configure an external Spark engine (EMR) to read and write UC managed tables
- Compare this pattern with the centralized processing pattern in Section 3
