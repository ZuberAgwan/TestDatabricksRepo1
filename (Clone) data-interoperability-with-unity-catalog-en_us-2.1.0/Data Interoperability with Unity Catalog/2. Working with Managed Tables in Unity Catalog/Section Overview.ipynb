# 2. Working with Managed Tables in Unity Catalog - Overview

This section covers Unity Catalog governance, managed table types, Delta and Iceberg table formats, performance optimizations, and how to provide external access to managed tables.

## Learning Objectives

By the end of this section, you will be able to:
- Understand the three-level namespace and governance features in Unity Catalog
- Compare table types in Unity Catalog (managed, external, foreign)
- Optimize performance with automated features for managed tables
- Choose between managed Delta and managed Iceberg for a given workload
- Create Delta and Iceberg tables in Databricks (including Iceberg v3)
- Access Iceberg tables using UniForm
- Set up external access to UC
- Understand the role of the Iceberg REST catalog
