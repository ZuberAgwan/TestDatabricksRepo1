# 2.1 Lecture: Interoperability and Performance Optimization with Managed Tables in Unity Catalog

This lesson covers how Unity Catalog organizes assets through its three-level namespace, the table types and governance features that apply to managed tables, the performance optimizations that make managed tables the recommended default, and how to choose between and create managed Delta and managed Iceberg tables (including Iceberg v3).

## Learning Objectives

By the end of this lesson, you will be able to:
- Describe the three-level namespace and its benefits
- Identify UC governance features including FGAC, lineage, and audit logging
- Compare managed, external, and foreign table types
- Understand performance optimizations available for managed tables
- Choose between managed Delta and managed Iceberg for a given workload
- Create managed UC managed tables with UniForm enabled for Iceberg readers
- Create managed Iceberg tables at format version 3
- Identify the table properties that control v3 features

## C. Unity Catalog Table Types

| Table Type | Description | Supported Formats | Best For |
|-----------|-------------|-------------------|----------|
| **Managed Tables** | Native storage managed by UC. Full governance, optimized performance, automatic optimizations | `DELTA`, `ICEBERG` | Most use cases (recommended default) |
| **External Tables** | References data stored outside UC managed storage. UC manages metadata while data remains external | `DELTA`, `ICEBERG`, `AVRO`, `CSV`, `JSON`, `ORC`, `PARQUET`, `TEXT` | Data that cannot be moved |
| **Foreign Tables** | Federation to tables in other metastores (HMS, Glue). Migration path from legacy systems | Varies by source | Legacy integration, migration |

| Choose this... | When... |
|---|---|
| **Managed Delta** | (__*Default*__) Databricks (and Delta-aware tools like Fabric, Spark, Trino) is the primary reader; no external Iceberg clients required |
| **Managed Delta + UniForm** | You want Delta as the primary format but external Iceberg clients also need to read the table |
| **Managed Iceberg (v3)** | An external Iceberg engine is the primary writer or reader, or your organization has standardized on Iceberg |

## Key Takeaways

- The **three-level namespace** (Catalog &gt; Schema &gt; Table) enables clean organization and access control
- **Managed tables** are the recommended default, offering full governance and automatic optimizations
- **External tables** are for data that can't be moved; **foreign tables** bridge legacy systems
- **Predictive Optimization** delivers AI-powered performance improvements with zero manual effort
- Both **managed Delta** and **managed Iceberg** tables get full UC governance and the same optimization features
- Use **Delta + UniForm** when Databricks is the primary writer and external Iceberg clients also need to read
- Use **managed Iceberg (v3)** when an external Iceberg engine is the primary writer/reader
- The `format-version=3` and `delta.enableIcebergCompatV3` properties are the v3 entry points
