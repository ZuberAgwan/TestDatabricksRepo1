# 3. Centralized Data Processing with External Analytics - Overview

This section covers the first major interoperability use case: using Databricks as the centralized processing hub while letting external analytics engines (Snowflake, PyIceberg, and any other Iceberg-aware engine) read UC-governed tables through the Iceberg REST Catalog. The Snowflake demo shows the cross-platform analytics pattern; the lab walks through enabling Iceberg V2 and V3 reads on Delta, comparing deletion-vector behaviour, and verifying external access with PyIceberg.

## Learning Objectives

By the end of this section, you will be able to:
- Describe the centralized data processing pattern with external analytics consumers
- Configure Snowflake to query UC managed tables via the Iceberg REST API
- Enable Iceberg V2 and V3 reads on existing UC managed tables and explain the deletion-vector implications
- Recognize the difference between enabled-Delta and managed Iceberg via table properties
- Use PyIceberg to confirm which UC table types are visible to external Iceberg engines
