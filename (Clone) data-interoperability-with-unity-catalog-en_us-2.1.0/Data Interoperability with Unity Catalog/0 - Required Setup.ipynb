    <div style="display: flex; align-items: flex-start; gap: 12px;">
        <div>
            <strong style="color: #37474f; font-size: 1.1em;">Step 2: Create the Course Schema</strong>
            <p style="margin: 8px 0 0 0; color: #333;">All demos and labs share a single schema, <code>data_interoperability_tpcds</code>, in your user catalog. This is dropped and recreated each time you run this notebook.</p>
        </div>

```sql
-- Drop and recreate the course schema
DROP SCHEMA IF EXISTS data_interoperability_tpcds CASCADE;
CREATE SCHEMA data_interoperability_tpcds;
USE SCHEMA data_interoperability_tpcds;

SELECT
  current_catalog() AS catalog,
  current_schema()  AS schema;
```

<div style="font-size: 1em; border-left: 4px solid #607d8b; background: #eceff1; padding: 16px 20px; border-radius: 4px; margin: 16px 0;">
    <div style="display: flex; align-items: flex-start; gap: 12px;">
        <div>
            <strong style="color: #37474f; font-size: 1.1em;">Step 3: Create Dimension Views</strong>
            <p style="margin: 8px 0 0 0; color: #333;">The dimension tables we join against (<code>date_dim</code>, <code>item</code>, <code>customer</code>, <code>store</code>) are exposed as views over <code>samples.tpcds_sf1000.*</code>. They are small enough that a view is fine and avoids unnecessary materialization.</p>
        </div>
    </div>
</div>

```sql
CREATE OR REPLACE VIEW date_dim AS SELECT * FROM samples.tpcds_sf1000.date_dim;
CREATE OR REPLACE VIEW item     AS SELECT * FROM samples.tpcds_sf1000.item;
CREATE OR REPLACE VIEW customer AS SELECT * FROM samples.tpcds_sf1000.customer;
CREATE OR REPLACE VIEW store    AS SELECT * FROM samples.tpcds_sf1000.store;

SELECT 'date_dim' AS view_name, COUNT(*) AS rows FROM date_dim
UNION ALL SELECT 'item',     COUNT(*) FROM item
UNION ALL SELECT 'customer', COUNT(*) FROM customer
UNION ALL SELECT 'store',    COUNT(*) FROM store;
```
