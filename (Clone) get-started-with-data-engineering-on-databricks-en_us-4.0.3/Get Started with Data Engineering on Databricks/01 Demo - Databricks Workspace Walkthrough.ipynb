### 4. Find data in the catalog

All governed data in Databricks lives in **Unity Catalog**, organized in a three-level hierarchy: **catalog → schema → table**. Let's find a sample table that's available in every Databricks workspace.

#### Browse to the table

1. In the left sidebar, click **Catalog** to open Catalog Explorer.
1. Find and expand the **`samples`** catalog.
1. Expand the **`nyctaxi`** schema.
1. Click the **`trips`** table.
1. On the right panel, look at the **Overview** tab to see the column list, then click the **Sample Data** tab to preview a few rows.

You're looking at NYC taxi trip data: pickup and dropoff times, fare amounts, distances, and zip codes.

#### Query that same table from this notebook

Run the cell below to query the table you just previewed.

```python
%sql
SELECT *
FROM samples.nyctaxi.trips
LIMIT 10;
```

You're now seeing the same data you previewed in Catalog Explorer, just queried directly in the notebook. This is the loop you'll use throughout the course: **find data in the catalog, then work with it in a notebook.**

### 5. Notebook basics: cells, languages, and Genie Code
Databricks notebooks support multiple languages in the same notebook. The cell below is Python. It counts the rows in the sample table and prints the result.

```python
trip_count = spark.sql("SELECT COUNT(*) AS total FROM samples.nyctaxi.trips").collect()[0]["total"]
print(f"There are {trip_count:,} trips in the sample dataset.")
```
